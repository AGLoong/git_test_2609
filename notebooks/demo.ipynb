这是项目的 Jupyter Notebook 目录，用于数据分析、模型探索和结果可视化。

```python
import numpy as np
import matplotlib.pyplot as plt

# 示例：生成并可视化数据分布
np.random.seed(42)
data = np.random.randn(1000)

fig, ax = plt.subplots(figsize=(8, 4))
ax.hist(data, bins=50, color="steelblue", edgecolor="white", alpha=0.8)
ax.set_title("Data Distribution Demo", fontsize=14)
ax.set_xlabel("Value")
ax.set_ylabel("Frequency")
plt.tight_layout()
plt.show()
```
